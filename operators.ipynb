```python
# Arithmetic Operators: Used for mathematical calculations
# Comparison Operators: Used to compare values → returns True/False
# Logical Operators   : Used to combine conditions
# Assignment Operators: Used to assign values
# Bitwise Operators   : Work on binary (bits)
# Membership Operators: Check if value exists in a sequence
# Identity Operators  : Check memory location (object identity)
```

```python
# Arithmetic Operators: Used for mathematical calculations

# | Name           | Symbol | Usage    | What it does             |
# | -------------- | ------ | -------- | ------------------------ |
# | Addition       | `+`    | `a + b`  | Adds values              |
# | Subtraction    | `-`    | `a - b`  | Subtracts values         |
# | Multiplication | `*`    | `a * b`  | Multiplies               |
# | Division       | `/`    | `a / b`  | Returns float division   |
# | Floor Division | `//`   | `a // b` | Returns integer division |
# | Modulus        | `%`    | `a % b`  | Remainder                |
# | Exponent       | `**`   | `a ** b` | Power                    |

# BODMAS is a rule that tells you the correct order to solve a mathematical expression.
# B → Brackets, Paranthasis
# O → Orders (powers, roots)
# DM → Division, Multiplication
# As → Addition, Subtraction

# ()
# ** (², √)
# * / // %
# + -
```

```python
# Comparison Operators: Used to compare values
# returns True/False

# | Name          | Symbol | Usage    | What it does          |
# | ------------- | ------ | -------- | --------------------- |
# | Equal         | `==`   | `a == b` | Checks value equality |
# | Not Equal     | `!=`   | `a != b` | Checks not equal      |
# | Greater Than  | `>`    | `a > b`  | Greater               |
# | Less Than     | `<`    | `a < b`  | Less                  |
# | Greater Equal | `>=`   | `a >= b` | Greater or equal      |
# | Less Equal    | `<=`   | `a <= b` | Less or equal         |

# True → 1
# False → 0
```

```python
# Logical Operators: Used to combine conditions
# returns True/False

# | Name | Symbol | Usage     | What it does      |
# | ---- | ------ | --------- | ----------------- |
# | AND  | `and`  | `a and b` | True if both True |
# | OR   | `or`   | `a or b`  | True if one True  |
# | NOT  | `not`  | `not a`   | Reverse result    |

# and
# | A | B | Result |
# | - | - | ------ |
# | T | T | T      |
# | T | F | F      |
# | F | T | F      |
# | F | F | F      |

# or
# | A | B | Result |
# | - | - | ------ |
# | T | T | T      |
# | T | F | T      |
# | F | T | T      |
# | F | F | F      |

# or
# | A | B | Result |
# | - | - | ------ |
# | T | T | T      |
# | T | F | T      |
# | F | T | T      |
# | F | F | F      |

# First condition is False
# Second condition is NOT executed → avoids error

# and → stops if False
# or → stops if True

# and → returns first False or last value
# or → returns first True value

# Order: not, and, or

# Use & instead of and in Pandas
```

```python
# Assignment Operators: Used to assign values

# | Name         | Symbol | Usage     | What it does |
# | ------------ | ------ | --------- | ------------ |
# | Assign       | `=`    | `a = 5`   | Assign value |
# | Add Assign   | `+=`   | `a += 2`  | `a = a + 2`  |
# | Sub Assign   | `-=`   | `a -= 2`  | `a = a - 2`  |
# | Mul Assign   | `*=`   | `a *= 2`  | `a = a * 2`  |
# | Div Assign   | `/=`   | `a /= 2`  | `a = a / 2`  |
# | Mod Assign   | `%=`   | `a %= 2`  | `a = a % 2`  |
# | Floor Assign | `//=`  | `a //= 2` | `a = a // 2` |
# | Power Assign | `**=`  | `a **= 2` | `a = a ** 2` |
```

```python
# Bitwise operators work on binary (0/1) representation of numbers.
# Every number in Python is stored in binary internally

# | Name        | Symbol | Usage    | What it does                           |
# | ----------- | ------ | -------- | -------------------------------------- |
# | AND         | `&`    | `a & b`  | 1 only if both bits are 1              |
# | OR          | `|`    | `a | b`  | 1 if any bit is 1                      |
# | XOR         | `^`    | `a ^ b`  | 1 if bits are different                |
# | NOT         | `~`    | `~a`     | Flips bits (1 → 0, 0 → 1)              |
# | Left Shift  | `<<`   | `a << n` | Shifts bits to left → multiplies by 2  |
# | Right Shift | `>>`   | `a >> n` | Shifts bits to right → divides by 2    |
```

```python
# Membership Operators: Check if value exists in a sequence
# returns True/False

# | Name   | Symbol   | Usage           | What it does         |
# | ------ | -------- | --------------- | -------------------- |
# | In     | `in`     | `a in list`     | Exists in collection |
# | Not In | `not in` | `a not in list` | Not exists           |

# performance

# | Data Type | Time Complexity |
# | --------- | --------------- |
# | List      | O(n)            |
# | Tuple     | O(n)            |
# | Set       | O(1) ⚡          |
# | Dict      | O(1) ⚡          |

# Use set for fast lookup
```

```python
# Identity Operators: Check memory location (object identity)
# returns True/False

# | Name   | Symbol   | Usage        | What it does     |
# | ------ | -------- | ------------ | ---------------- |
# | Is     | `is`     | `a is b`     | Same memory      |
# | Is Not | `is not` | `a is not b` | Different memory |
```

```python
# Employee data
salary = 60000
experience = 4
dept = "IT"
valid_depts = ["IT", "HR", "Finance"]

bonus_percent = 10
permissions = 5   # binary 101 (READ + EXEC)

data = None

# 1️⃣ Arithmetic Operators
bonus = salary * bonus_percent / 100 

# 2️⃣ Comparison Operators
is_high_salary = salary > 50000 

# 3️⃣ Logical Operators
is_eligible = is_high_salary and experience >= 3 

# 4️⃣ Membership Operators
is_valid_dept = dept in valid_depts   

# 5️⃣ Identity Operators
if data is None:                
    data = "No Data Available"

# 6️⃣ Assignment Operators
salary += bonus                  

# 7️⃣ Bitwise Operators (permissions)
READ = 1    # 001
WRITE = 2   # 010
EXEC = 4    # 100

has_read = permissions & READ       
has_write = permissions & WRITE      
has_exec = permissions & EXEC       

# Final Output
print("Updated Salary:", salary)
print("Eligible:", is_eligible)
print("Valid Dept:", is_valid_dept)
print("Permissions - Read:", bool(has_read), 
      "Write:", bool(has_write), 
      "Exec:", bool(has_exec))
```
