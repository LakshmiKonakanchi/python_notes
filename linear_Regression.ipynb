### Linear Regression Notes

#### What is Linear Regression?

Linear Regression is a **Supervised Machine Learning algorithm** used to predict a **continuous numerical value** based on one or more independent variables.

**Examples**
- Predicting house prices
- Predicting sales revenue
- Predicting student marks
- Predicting temperature

---

### Types of Linear Regression

#### Simple Linear Regression

- Uses **one independent variable (X)**
- Predicts **one dependent variable (Y)**

**Example:** Predicting salary based on years of experience.

**Equation**

$$
\hat{Y}=b_0+b_1X
$$

#### Multiple Linear Regression

- Uses **two or more independent variables**
- Predicts one dependent variable.

**Example:** Predicting house price using:
- Area
- Number of bedrooms
- Age of the house

**Equation**

$$
\hat{Y}=b_0+b_1X_1+b_2X_2+\cdots+b_nX_n
$$

---

### Terminology

| Term | Meaning |
|------|---------|
| X | Independent Variable (Feature) |
| Y | Dependent Variable (Target) |
| Ŷ | Predicted Value |
| b₀ | Intercept |
| b₁ | Slope/Coefficient |
| Residual | Prediction Error |

---

### Linear Regression Equation

#### Simple Linear Regression

$$
\hat{Y}=b_0+b_1X
$$

Where:

- **Ŷ** = Predicted value
- **b₀** = Intercept
- **b₁** = Slope
- **X** = Independent variable

#### Understanding the Equation

**Intercept (b₀)**

The value of **Y when X = 0**.

**Slope (b₁)**

Shows how much **Y changes** for every **1-unit increase in X**.

---

#### Goal of Linear Regression

The goal is to find the **Best Fit Line** that minimizes prediction errors.

The algorithm uses the **Least Squares Method**.

---

#### Best Fit Line

A best-fit line is the line that has the **minimum total error** between actual values and predicted values.

$$
\hat{Y}=b_0+b_1X
$$

---

#### Residual (Error)

Residual is the difference between the actual value and the predicted value.

**Formula**

$$
\text{Residual}=Y-\hat{Y}
$$

Where:

- **Y** = Actual value
- **Ŷ** = Predicted value

---

#### Sum of Squared Errors (SSE)

Instead of adding errors directly, Linear Regression squares every error.

**Formula**

$$
SSE=\sum(Y-\hat{Y})^2
$$

**Objective**

> Minimize SSE

Smaller SSE means a better model.

---

#### Mean Squared Error (MSE)

Average of squared errors.

**Formula**

$$
MSE=\frac{1}{n}\sum(Y-\hat{Y})^2
$$

Where:

- **n** = Number of observations

Lower MSE indicates better performance.

---

#### Root Mean Squared Error (RMSE)

Square root of MSE.

**Formula**

$$
RMSE=\sqrt{MSE}
$$

RMSE has the same unit as the target variable, making it easier to interpret.

---

#### Mean Absolute Error (MAE)

Average of the absolute errors.

**Formula**

$$
MAE=\frac{1}{n}\sum|Y-\hat{Y}|
$$

MAE is less sensitive to outliers than MSE.

---

#### R-Squared (Coefficient of Determination)

Measures how well the model explains the variance in the target variable.

**Formula**

$$
R^2=1-\frac{SS_{res}}{SS_{tot}}
$$

Where:

$$
SS_{res}=\sum(Y-\hat{Y})^2
$$

$$
SS_{tot}=\sum(Y-\bar{Y})^2
$$

**Interpretation**

- **R² = 1** → Perfect prediction
- **R² = 0** → Model explains no variance
- Higher R² is generally better.

---

#### Worked Example

Regression equation:

$$
\hat{Y}=5+2X
$$

If:

$$
X=10
$$

Then:

$$
\hat{Y}=5+2(10)=25
$$

**Predicted Value = 25**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error, root_mean_squared_error
```

```python
df = pd.read_csv('/home/madhu/Anvisoft_backup/python_projects/datasets/Advertising Budget and Sales.csv')
```

```python
x = df[['TV Ad Budget ($)', 'Radio Ad Budget ($)', 'Newspaper Ad Budget ($)']] #100%
y = df['Sales ($)']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
LR = LinearRegression()
LR.fit(x_train, y_train)
y_pred = LR.predict(x_test) # y--> actual sales , y_pred --> predicted sales
```

```python
y_pred
```

```python
y
```

```python
print("Mean Squared Error : ", mean_squared_error(y_test, y_pred))
print("R^2 Score : ", r2_score(y_test, y_pred))
print("Mean Absolute Error : ", mean_absolute_error(y_test, y_pred))
print("Root Mean Squared Error : ", root_mean_squared_error(y_test, y_pred))
```

```python
y_pred_train = LR.predict(x_train)

print("Mean Squared Error (MSE):", mean_squared_error(y_train, y_pred_train))
print("R^2 Score (R-squared):", r2_score(y_train, y_pred_train))
print("Mean Absolute Error (MAE):", mean_absolute_error(y_train, y_pred_train))
print("Root Mean Squared Error (RMSE):", root_mean_squared_error(y_train, y_pred_train))
```
