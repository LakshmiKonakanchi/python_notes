```python
# difference between append() and extend() in Python lists

a = [1, 2]

a.append([3, 4])
print(a)  

a = [1, 2]
a.extend([3, 4])
print(a)  

# append() → adds entire object as single element
# extend() → adds elements individually
```

```python
list1 = ['id', 'name', 'salary', 'age', 'country']
list2 = [101, 'Lakshmi', 50000]

from itertools import zip_longest
print(list(zip_longest(list1, list2, fillvalue=None)))

list_zip= (list(zip(list1,list2)))
dict_zip= (dict(zip(list1,list2)))
set_zip= (set(zip(list1,list2)))
tuple_zip= (tuple(zip(list1,list2)))

print(list_zip, dict_zip, set_zip, tuple_zip, sep='\n')

list1, list2 = zip(*list_zip)
print(list1, list2, sep='\n')


```
