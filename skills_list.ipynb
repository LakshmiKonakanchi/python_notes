Snowflake

```python
# Snowflake is a cloud-native Data Warehouse.
# It runs on: AWS, Microsoft Azure, Google Cloud

# Traditional Warehouse: Storage + Compute Together

# But in Snowflake:   Storage
                    #    |
                    #    |
                    # Compute Cluster 1
                    # Compute Cluster 2
                    # Compute Cluster 3

# Storage and Compute are separated in Snowflake's biggest advantage.

# If workload increases:  Small Warehouse
                        #       ↓
                        # Medium Warehouse
                        #       ↓
                        # Large Warehouse

# Snowflake Architecture: Database Storage Layer
#                         Cloud Services Layer
#                         Virtual Warehouse Layer

# Virtual Warehouse:Independent compute cluster, used for query execution.

# Common Tools: Python, Airflow, dbt, Power BI, Tableau, Spark
```
