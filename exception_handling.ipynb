```python
# Exception handling is how Python deals with runtime errors without crashing your program.

# try:
#     risky code
# except:
#     handle error

try:
    x = int("abc")
except:
    print("Conversion failed")
```

```python
# try:
#     risky code
# except ExceptionType:
#     handle specific error
# else:
#     runs if NO error
# finally:
#     always runs

try:
    x = int("10.5")
except ValueError:
    print("Invalid conversion")
else:
    print("Success:", x)
finally:
    print("Execution completed")
```

```python
# Common types of Exceptions:

# | Exception         | When it happens        |
# | ----------------- | ---------------------- |
# | ValueError        | Wrong value type       |
# | TypeError         | Wrong data type        |
# | IndexError        | Invalid index          |
# | KeyError          | Missing dictionary key |
# | FileNotFoundError | File missing           |
# | ZeroDivisionError | Divide by zero         |
```

```python
data = {"name": "Lakshmi"}

try:
    print(data["age"])
except KeyError:
    print("Key not found")

try:
    x = int("abc")
except (ValueError, TypeError):
    print("Error occurred")    
```

```python
# Using as (Very Important)
try:
    x = 10 / 0
except ZeroDivisionError as e:
    print("Error:", e)
```

```python
try:
    x = 10/0
except Exception as e:
    print("Type:", type(e).__name__)
    print("Message:", e)
    print()
    print(f"Type: {type(e).__name__}\nMessage: {e}")
```

```python
# Use sep → between values
# Use end → after output
# Use empty print() → for readability
# After every print there is (,end='\n') defaulty, invisible
# print("Hello",end='\n') = print("Hello")

a = "Lakshmi"
b = "Narasimha"
print(a, b, sep='\n', end='\n\n')
print(f"{a}\n{b}")
print(f"\n{a}\n{b}")
print()
print("Lakshmi ", end="")
print("Narasimha")
```
