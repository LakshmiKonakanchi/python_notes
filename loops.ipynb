```python
# for loop → used when iteration count is known (uses iterator internally)
# while loop → used when condition-based looping is required


# for loop for readability and safety.
# while when iteration count is unknown or depends on runtime conditions.

# for loop is iterator-based, so:
#       No risk of forgetting increment
#       Less chance of infinite loop
# while loop:
#       Manual control → more flexible
#       But prone to logical errors (missing updates)

# Performance:
# for is usually faster for sequences (optimized in Python)
# while gives more control but risk of infinite loops

# for : Used when you know how many times or iterate over items
# when: Runs as long as a condition is True
# break – stops the loop completely
# continue – skips current iteration
# pass – does nothing (placeholder)
# nested loop: A loop inside another loop
# loop with else: else runs only if loop did not break
```

```python
# | Loop | Increment Position | Output Type            |
# | ---- | ------------------ | ---------------------- |
# | 1    | before print       | 1 → 11                 |
# | 2    | after print        | 0 → 10                 |
# | 3    | before & after     | odd numbers            |
# | 4    | double before      | even numbers (start 2) |
# | 5    | double after       | even numbers (start 0) |

# | Case                   | Rule                       |
# | ---------------------- | -------------------------- |
# | Print after increment  | starts from next number    |
# | Print before increment | starts from current number |
# | 2 increments           | skips numbers              |
# | increment before print | affects output             |
# | increment after print  | affects next loop          |

i = 0
while i<=10:
    i+=1
    print(i, end=" ")

# Starts at 0 → becomes 1 → prints
# Continues till i becomes 11 (because condition checked before increment)
# Increment happens before printing
print() 
# | Step | i (before) | Condition (i<=10) | i after `i+=1` | Printed |
# | ---- | ---------- | ----------------- | -------------- | ------- |
# | 1    | 0          | True              | 1              | 1       |
# | 2    | 1          | True              | 2              | 2       |
# | 3    | 2          | True              | 3              | 3       |
# | ...  | ...        | ...               | ...            | ...     |
# | 11   | 10         | True              | 11             | 11      |
# | 12   | 11         | False             | —              | —       |
 
i = 0
while i<=10:
    print(i, end=" ")
    i+=1

# Starts at 0 → prints
# Continues till i becomes 10 (because condition checked after increment)
# printing happens before increment
print()  
# | Step | i (before) | Condition | Printed | i after |
# | ---- | ---------- | --------- | ------- | ------- |
# | 1    | 0          | True      | 0       | 1       |
# | 2    | 1          | True      | 1       | 2       |
# | ...  | ...        | ...       | ...     | ...     |
# | 11   | 10         | True      | 10      | 11      |
# | 12   | 11         | False     | —       | —       |

i = 0
while i<=10:
    i+=1
    print(i, end=" ")
    i+=1

# 0 → 1 (print)
# 1 → 2 → next loop
# 2 → 3 (print)
# 3 → 4 → next loop
# Pattern: odd numbers
print()
# | Step | i (start) | after +1 | Printed | after +1 again |
# | ---- | --------- | -------- | ------- | -------------- |
# | 1    | 0         | 1        | 1       | 2              |
# | 2    | 2         | 3        | 3       | 4              |
# | 3    | 4         | 5        | 5       | 6              |
# | 4    | 6         | 7        | 7       | 8              |
# | 5    | 8         | 9        | 9       | 10             |
# | 6    | 10        | 11       | 11      | 12             |

i = 0
while i<=10:
    i+=1
    i+=1
    print(i, end=" ")

# 1 → 2 (print)
# 2 → 3 → next loop
# 3 → 4 (print)
# 4 → 5 → next loop
# Pattern: even numbers
print()
# | Step | i (start) | after +2 | Printed |
# | ---- | --------- | -------- | ------- |
# | 1    | 0         | 2        | 2       |
# | 2    | 2         | 4        | 4       |
# | 3    | 4         | 6        | 6       |
# | 4    | 6         | 8        | 8       |
# | 5    | 8         | 10       | 10      |
# | 6    | 10        | 12       | 12      |

i = 0
while i<=10:
    print(i, end=" ")
    i+=1
    i+=1

# Pattern: even numbers starting from 0
# | Step | i (start) | Printed | after +2 |
# | ---- | --------- | ------- | -------- |
# | 1    | 0         | 0       | 2        |
# | 2    | 2         | 2       | 4        |
# | 3    | 4         | 4       | 6        |
# | 4    | 6         | 6       | 8        |
# | 5    | 8         | 8       | 10       |
# | 6    | 10        | 10      | 12       |
```

```python
# Used in:Accumulation problems, Financial calculations
# It keeps adding numbers (1,2,3,4,...)
# Stops when total becomes ≥ 50
# Loop condition: total < 50, but it does not stop exactly at 50, so it stops after crossing 50. That’s why result = 55
# 1 + 2 + 3 + ... + n ≥ 50

total = 0
i = 1
while total < 50:
    total += i
    i += 1
print(total)        # loop ends after 50

total = 0
i = 1
while total + i <= 50:
    total += i
    i += 1
print(total)        # loop ends before 50
```

```python
# % 10 → get last digit
# // 10 → remove last digit

# | Step | num       | digit (`num % 10`) | rev calculation | rev       |
# | ---- | --------- | ------------------ | --------------- | --------- |
# | 1    | 123456789 | 9                  | 0*10 + 9        | 9         |
# | 2    | 12345678  | 8                  | 9*10 + 8        | 98        |
# | 3    | 1234567   | 7                  | 98*10 + 7       | 987       |
# | 4    | 123456    | 6                  | 987*10 + 6      | 9876      |
# | 5    | 12345     | 5                  | 9876*10 + 5     | 98765     |
# | 6    | 1234      | 4                  | 98765*10 + 4    | 987654    |
# | 7    | 123       | 3                  | 987654*10 + 3   | 9876543   |
# | 8    | 12        | 2                  | 9876543*10 + 2  | 98765432  |
# | 9    | 1         | 1                  | 98765432*10 + 1 | 987654321 |

# taking digits from right side
# building new number from left side

num = 123456789
rev = 0

while num > 0:
    digit = num % 10
    rev = rev * 10 + digit
    num //= 10
print(rev)

num = 123456789
rev = 0

rev_str = int(str(num)[::-1])
print(rev_str)
```

```python
import os
import time

start_time = time.time()
timeout = 30  # seconds

while not os.path.exists("data.csv"):
    if time.time() - start_time > timeout:
        print("File not found after 30 seconds. Exiting...")
        break
    
    print("Waiting for file...")
    time.sleep(2)   # wait 2 seconds before checking again

# Optional: check if file found
if os.path.exists("data.csv"):
    print("File found!")

# time.time() → current time in seconds
# start_time → when loop started
# time.time() - start_time → elapsed time
# time.sleep(2) → prevents continuous checking (CPU friendly)
```

```python
# Always add sleep → avoids high CPU usage
# Use timeout → prevents infinite waiting
# Used in: ETL pipelines, Airflow jobs, Data ingestion systems

# time.time() → wall-clock time (can change if system time changes)
# time.perf_counter() → high-resolution timer (best for measuring durations)
# For timeouts, perf_counter() is more reliable
```

```python
def wait_for_file(filepath, timeout=30, interval=2):
    import os, time
    
    start = time.time()
    
    while not os.path.exists(filepath):
        if time.time() - start > timeout:
            return False
        time.sleep(interval)
    
    return True


if wait_for_file("data.csv"):
    print("File found!")
else:
    print("Timeout reached.")
```
