```python
class bank:
    def __init__(self,acc_no,acc_name,balance):
        self.acc_no = acc_no 
        self.acc_name = acc_name   #attribute
        self.__balance = balance   # private attribute

    def deposit(self,amount):
        self.__balance += amount

    def get_balance(self):
        print(f'acc_n0:{self.acc_no},acc_name:{self.acc_name}')
        return self.__balance
```

```python
b1= bank(2000,"madhu",10000)
```

```python
print(b1.get_balance())
```

```python
class bank_b :
    def __init__(self,balance):
        self.__balance = balance   # private attribute

    def deposit(self,amount):
        self.__balance += amount

    def get_balance_b(self):
        return self.__balance
```

```python
from abc import ABC, abstractmethod

class shape(ABC):
    @abstractmethod
    def area(self):
        pass
class circle(shape):
    def __init__(self,r):
        self.r = r
    def  area(self):
        return 3.14 * self.r * self.r
```

```python
c = circle(5)
print(c.area())
```

```python
class emp:
    com = "google"
    def __init__(self,name,salary):
        self.name = name
        self.salary = salary
    @classmethod
    def change_company(cls,new_com):
        cls.com = new_com
    @staticmethod
    def office_time():
        return "(9 AM to 6 PM)"
```

```python
e1 = emp("madhu",100000)
```

```python
e1.com
```

```python
e1.office_time()
```

```python
emp.change_company("openai")
e1.company
```

```python
# updating object data using methods

class account:
    def __init__(self,name,balance):
        self.name = name
        self.balance = balance

    def add_money(self,amount):
        self.balance += amount
```

```python
# single vs mltilavel vs multple inheritance

class a:
    def amsg(self):
        print("class a")
class b(a):
    def bmsg(self):
        print("class b")
class c(b):
    def cmsg(self):
        print("class c")
```

```python
obj = c()
```

```python
obj.amsg()
obj.bmsg()
obj.cmsg()
```
