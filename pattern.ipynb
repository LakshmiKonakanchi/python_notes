```python
# right-angle triangle using stars.

n = int(input("Enter number: "))

for i in range(1, n + 1):
    for j in range(i):
        print("*", end=" ")
    print()

print()

for i in range(1, n + 1):
    print("* " * i)

# * " * i repeats the string automatically.

print()

for i in range(1, n+1): print("* "*i)
```

```python
# inverted right-angle triangle.

n = int(input("Enter number: "))

for i in range(n, 0, -1):
    for j in range(i):
        print("*", end=" ")
    print()

print()

for i in range(n, 0, -1):
    print("* " * i)

print()

for i in reversed(range(1, n + 1)):
    print("* " * i)

print()

for i in range(n, 0, -1): print("* " * i)

# range(n, 0, -1) → starts from n, decreases by 1.
# * " * i → prints i stars in each row.
```

```python
# number triangle.

n = int(input("Enter number: "))

for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()

print()

for i in range(1, n + 1):
    print(*range(1, i + 1))

# range() generates numbers.
# * unpacks values.
# print() automatically adds spaces.

print()

for i in range(1, n + 1):
    print(" ".join(map(str, range(1, i + 1))))

print()

for i in range(1, n+1): print(*range(1, i+1))

# For pattern programs, try replacing nested loops with:

# string multiplication → for star patterns
# range() + unpacking (*) → for number patterns
# join() → when formatting output
```

```python
# reverse (inverted) number triangle.

n = int(input("Enter number: "))

for i in range(n, 0, -1):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()

print()

for i in range(n, 0, -1):
    print(*range(1, i + 1))

# range(1, i + 1) → generates numbers
# * (unpacking) → prints with spaces automatically

print()

for i in range(n, 0, -1): print(*range(1, i+1))

print()

for i in range(n, 0, -1):
    print(" ".join(map(str, range(1, i + 1))))

# For reverse number patterns:
# Outer loop → decreasing (n → 1)
# Inner values → increasing (1 → i)

# Formula:
# Rows decrease → n to 1
# Numbers increase → 1 to i
```

```python
# Floyd’s (continuous number) triangle.

n = int(input("Enter number: "))
num = 1

for i in range(1, n + 1):
    for j in range(i):
        print(num, end=" ")
        num += 1
    print()

print()

for i in range(1, n + 1):
    print(*range(num, num + i))
    num += i

# range(num, num + i) → prints i numbers in a row.
# num += i → updates starting number for next row.

print()

nums = iter(range(2, n * (n + 1) // 2 + 2))
for i in range(1, n + 1):
    print(*(next(nums) for _ in range(i)))
```

```python
# pyramid (right-aligned) star pattern.

n = int(input("Enter number: "))

for i in range(1, n + 1):
    for j in range(n - i):  # spaces
        print(" ", end="")
    for k in range(i):      # stars
        print("* ", end="")
    print()

print()

for i in range(1, n + 1):
    print(" " * (n - i) + "* " * i)

# " " * (n - i) → creates left spaces.
# "* " * i → prints stars.

print()

for i in range(1, n+1): print(" "*(n-i) + "* "*i)

# Spaces decrease → n - i
# Stars increase → i
# This trick works for most pyramid patterns

print()

for i in range(1, n + 1):
    print(("* " * i).center(n * 2))
```

```python
# reverse (inverted) star triangle (right-aligned).

n = int(input("Enter number: "))

for i in range(n, 0, -1):
    for j in range(n - i):
        print(" ", end="")
    for k in range(i):
        print("* ", end="")
    print()

print()

for i in range(n, 0, -1):
    print(" " * (n - i) + "* " * i)

# " " * (n - i) → increasing spaces
# "* " * i → decreasing stars

print()

for i in range(n, 0, -1): print(" "*(n-i) + "* "*i)

print()

for i in range(n, 0, -1):
    print(("* " * i).center(n * 2))

# For inverted patterns:
# Spaces increase → n - i
# Stars decrease → i

# This same logic works for:
# Inverted pyramids
# Number reverse patterns
# Alphabet reverse patterns
```

```python
# diamond pattern (Pyramid + Reverse Pyramid)

n = int(input("Enter number: "))

# upper part
for i in range(1, n + 1):

    for j in range(n - i):
        print(" ", end="")
    for k in range(i):
        print("* ", end="")
    print()

# lower part
for i in range(n - 1, 0, -1):

    for j in range(n - i):
        print(" ", end="")
    for k in range(i):
        print("* ", end="")
    print()

print()

for i in range(1, n + 1):
    print(" " * (n - i) + "* " * i)

for i in range(n - 1, 0, -1):
    print(" " * (n - i) + "* " * i)

print()

for i in list(range(1, n+1)) + list(range(n-1, 0, -1)):
    print(" " * (n - i) + "* " * i)

print()

for i in range(1, n * 2):
    stars = i if i <= n else 2 * n - i
    print(" " * (n - stars) + "* " * stars)

# diamond as: Upper triangle + Lower inverted triangle

# Spaces → n - i
# Stars → i
# Upper: 1 → n
# Lower: n-1 → 1

# This logic helps solve: Diamond, Hollow diamond, Butterfly pattern, Sandglass pattern
```

```python
# hollow square pattern.

n = int(input("Enter number: "))

for i in range(n):
    for j in range(n):
        if i == 0 or i == n-1 or j == 0 or j == n-1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()

print()

for i in range(n):
    print("* " * n if i in (0, n-1)
          else "* " + "  " * (n-2) + "* ")
    
print()

for i in range(n): print("* "*n if i in (0,n-1) else "* " + "  "*(n-2) + "* ")

# For hollow patterns, boundary condition is key:
# Top row    → i == 0
# Bottom row → i == n-1
# Left side  → j == 0
# Right side → j == n-1

# If boundary → print *
# Else → print space

# This same idea works for: Hollow rectangle, Hollow pyramid, Hollow diamond
```

```python
# butterfly star pattern.

n = int(input("Enter number: "))

# Upper part
for i in range(1, n + 1):

    for j in range(i):
        print("*", end="")    # Left stars

    for j in range(2 * (n - i)):
        print(" ", end="")    # Spaces

    for j in range(i):
        print("*", end="")    # Right stars

    print()

# Lower part
for i in range(n - 1, 0, -1):

    for j in range(i):
        print("*", end="")

    for j in range(2 * (n - i)):
        print(" ", end="")

    for j in range(i):
        print("*", end="")

    print()

print()

for i in range(1, n + 1):
    print("*" * i + " " * (2 * (n - i)) + "*" * i)

for i in range(n - 1, 0, -1):
    print("*" * i + " " * (2 * (n - i)) + "*" * i)

print()

for i in list(range(1, n+1)) + list(range(n-1, 0, -1)):
    print("*" * i + " " * (2 * (n - i)) + "*" * i)

print()

rows = list(range(1, n+1)) + list(range(n-1, 0, -1))

for i in rows:
    print("*"*i + " "*(2*(n-i)) + "*"*i)

# butterfly pattern as: Left Stars + Middle Spaces + Right Stars

# Formula:
# Stars  → i
# Spaces → 2 × (n - i)

# Pattern Flow:
# Upper  → 1 to n
# Lower  → n-1 to 1
```

```python
# hollow pyramid.

n = int(input("Enter number: "))

for i in range(n):

    # spaces
    for j in range(n - i - 1):
        print(" ", end="")

    # stars and hollow
    for j in range(2 * i + 1):

        if j == 0 or j == 2 * i or i == n - 1:
            print("*", end="")
        else:
            print(" ", end="")

    print()

# Border condition:
# j == 0
# j == 2*i
# i == n-1

print()

for i in range(n):
    spaces = " " * (n - i - 1)

    if i == n - 1:
        stars = "*" * (2 * i + 1)
    else:
        stars = "*" + " " * (2 * i - 1) + "*" if i > 0 else "*"

    print(spaces + stars)

print()

for i in range(n):
    print(
        " " * (n - i - 1),
        "*" if i == 0 else
        "*" * (2*i + 1) if i == n - 1 else
        "*" + " " * (2*i - 1) + "*",
        sep=""
    )

print()

for i in range(n):
    print(" "*(n-i-1) +
          ("*"*(2*i+1) if i==n-1
           else "*"+" "*(2*i-1)+"*" if i else "*"))
    
# hollow pyramid:
# Left spaces = n - i - 1
# Width       = 2*i + 1

# Boundary condition:
# First star → j == 0
# Last star  → j == 2*i
# Last row   → i == n-1 (full stars)
```

```python
# Pascal's triangle.

n = int(input("Enter number: "))

for i in range(n):

    num = 1

    for j in range(n - i):
        print(" ", end="")

    for j in range(i + 1):
        print(num, end=" ")

        num = num * (i - j) // (j + 1)

    print()

print()

for i in range(n):
    num = 1
    print(" " * (n - i), end="")

    for j in range(i + 1):
        print(num, end=" ")
        num = num * (i - j) // (j + 1)

    print()

print()

from math import comb

for i in range(n):
    print(" " * (n - i), end="")
    print(*(comb(i, j) for j in range(i + 1)))
    
# Interviewers often ask: Why use this formula instead of factorial?”

# Because:

# Factorials are slower (n!)
# Repeated calculation wastes time
# Your approach computes next value from previous one → optimized O(n²)
# Dry Run (i = 4)

# Numbers generated:

# 1
# 1×4/1 = 4
# 4×3/2 = 6
# 6×2/3 = 4
# 4×1/4 = 1
```

```python
# binary triangle (0,1).

n = int(input("Enter number: "))

for i in range(1, n + 1):

    for j in range(1, i + 1):

        if (i + j) % 2 == 0:
            print("1", end=" ")
        else:
            print("0", end=" ")

    print()

# Use odd/even: (i + j) % 2

print()

for i in range(1, n + 1):
    print(*[1 if (i + j) % 2 == 0 else 0 for j in range(1, i + 1)])

print()

for i in range(1, n + 1):
    for j in range(1, i + 1):
        print((i + j) % 2, end=" ")
    print()

# (i+j)mod2

# Logic:
# Even (0) → print 1
# Odd (1) → print 0

# Shortcut:
# Even sum → 1
# Odd sum  → 0

print()

for i in range(1, n + 1):
    print(*[(i + j + 1) % 2 for j in range(i)])

# Why use (i + j) % 2?
# Because it creates an alternating pattern automatically, avoiding extra variables or toggles.

print()

for i in range(1, n + 1):
    num = i % 2
    for _ in range(i):
        print(num, end=" ")
        num = 1 - num
    print()

```

```python
# alphabet triangle.

n = int(input("Enter number: "))

for i in range(1, n + 1):

    ch = 65

    for j in range(i):
        print(chr(ch), end=" ")
        ch += 1

    print()

print()

for i in range(1, n + 1):
    print(*[chr(65 + j) for j in range(i)])

print()

for i in range(1, n + 1):
    print(*map(chr, range(65, 65 + i)))

# ASCII values: A = 65, B = 66, C = 67 . . .

print()

for i in range(1, n + 1):
    print(*map(chr, range(ord('A'), ord('A') + i)))

print()

print(chr(65) )  # A
print(ord('A')) # 65

print()
for i in range(1, n+1): print(*map(chr, range(65, 65+i)))

# Pattern Logic:
# Row number = number of alphabets
# Start from A
# Increase character each time

# Instead use of remembering chr(65) for A use ord('A')

# Formula:
# Character = chr(65 + j)
# Character = chr(ord('A') + j)
```

```python
# reverse (inverted) alphabet triangle

n = int(input("Enter number: "))

for i in range(n, 0, -1):

    ch = 65

    for j in range(i):
        print(chr(ch), end=" ")
        ch += 1

    print()

print()

for i in range(n, 0, -1):
    print(*map(chr, range(65, 65 + i)))

print()

for i in range(n, 0, -1):
    print(*map(chr, range(ord('A'), ord('A') + i)))

# map(chr, range(...)) converts ASCII → characters.

print()

for i in range(n, 0, -1): print(*map(chr, range(65, 65+i)))

# For reverse alphabet pattern:

# Rows decrease → n → 1
# Characters always start from A
# Characters increase till row count

# Formula: Character = chr(ord('A') + j)

# Instead use of remembering chr(65) for A use ord('A')
```

```python
# hollow rectangle

rows = int(input("Enter rows: "))
cols = int(input("Enter columns: "))

for i in range(rows):
    for j in range(cols):

        if i == 0 or i == rows - 1 or j == 0 or j == cols - 1:
            print("*", end="")
        else:
            print(" ", end="")

    print()

print()

for i in range(rows):
    print("*" * cols if i in (0, rows - 1)
          else "*" + " " * (cols - 2) + "*")
    
print()

for i in range(rows):
    if i == 0 or i == rows - 1:
        print("*" * cols)
    else:
        print("*" + " " * (cols - 2) + "*")

print()

for i in range(rows): print("*"*cols if i in (0, rows-1) else "*" + " "*(cols-2) + "*")

# Boundary condition:

# Top row    → i == 0
# Bottom row → i == rows - 1
# Left side  → j == 0
# Right side → j == cols - 1

# If boundary → *
# Else → space

# This same logic is used for:
# Hollow square
# Hollow rectangle
# Hollow pyramid
# Hollow diamond

# If cols = 1, this can behave differently. Safer version:

for i in range(rows):
    if i in (0, rows - 1) or cols == 1:
        print("*" * cols)
    else:
        print("*" + " " * (cols - 2) + "*")
```

```python
# diamond/hourglass (sandglass) pattern.
# Inverted Pyramid + Normal Pyramid

# Formula:
# Spaces → n - i
# Stars → 2 * i - 1

# Pattern flow:
# Upper → n to 1
# Lower → 2 to n

# Why 2*i - 1 ?  2i−1
# It creates odd-number star counts:
# 9, 7, 5, 3, 1, 3, 5, 7, 9

n = int(input("Enter number: "))

# upper
for i in range(n, 0, -1):

    for j in range(n - i):
        print(" ", end="")
    for j in range(2 * i - 1):
        print("*", end="")
    print()

# lower
for i in range(2, n + 1):

    for j in range(n - i):
        print(" ", end="")
    for j in range(2 * i - 1):
        print("*", end="")
    print()

print()

for i in range(n, 0, -1):
    print(" " * (n - i) + "*" * (2 * i - 1))

for i in range(2, n + 1):
    print(" " * (n - i) + "*" * (2 * i - 1))

print()

rows = list(range(n, 0, -1)) + list(range(2, n + 1))

for i in rows:
    print(" " * (n - i) + "*" * (2 * i - 1))

print()

for i in list(range(n,0,-1))+list(range(2,n+1)):
    print(" "*(n-i) + "*"*(2*i-1))

print()

for i in [*range(n, 0, -1), *range(2, n + 1)]:
    print(" " * (n - i) + "*" * (2 * i - 1))
```

```python
# zig-zag pattern

n = int(input("Enter number: "))

for i in range(3):
    for j in range(n):

        if ((i + j) % 4 == 0) or (i == 1 and j % 4 == 0):
            print("*", end=" ")
        else:
            print(" ", end=" ")

    print()

print()

for i in range(3):
    for j in range(n):
        print("*" if (i + j) % 4 == 0 or (i == 1 and j % 4 == 0) else " ", end=" ")
    print()

print()

for i in range(3):
    for j in range(n):
        condition = (i + j) % 4 == 0 or (i == 1 and j % 4 == 0)

        print("*" if condition else " ", end=" ")
    print()

# Why does this work?
# The pattern uses mathematical positions instead of complicated logic.

# Condition 1:

# (i+j)mod4=0
# This creates the downward diagonal stars.

# Condition 2:

# i=1andjmod4=0
# This creates the middle upward curve stars.

# Visual Understanding
# For n = 15:

# Row 0 → *       *       *
# Row 1 →   *   *   *   *
# Row 2 →     *       *

# This creates a zig-zag / sine-wave look.

print()

for i in range(3):
    for j in range(n): print("*" if (i+j)%4==0 or (i==1 and j%4==0) else " ", end=" ")
    print()

# “Why % 4?”
# Because the pattern repeats every 4 columns.

# Pattern cycle:
# Column repeat → every 4 positions

# This is a good example of solving patterns using math instead of many loops/conditions.
```

```python
# palindrome number pyramid pattern

n = int(input("Enter number: "))

for i in range(1, n + 1):

    for j in range(n - i):
        print(" ", end="")    # spaces

    for j in range(i, 0, -1):
        print(j, end="")      # descending

    for j in range(2, i + 1):
        print(j, end="")      # ascending

    print()

print()

for i in range(1, n + 1):
    print(" " * (n - i), end="")
    print("".join(map(str, range(i, 0, -1))),
          "".join(map(str, range(2, i + 1))), sep="")
    
print()

for i in range(1, n + 1):
    left = "".join(map(str, range(i, 0, -1)))
    right = "".join(map(str, range(2, i + 1)))

    print(" " * (n - i) + left + right)

# Think of it in 3 parts: Spaces + Descending Numbers + Ascending Numbers

# Formula:
# Spaces → n - i
# Descending → i → 1
# Ascending → 2 → i

# Why ascending starts from 2?
# Because 1 is already printed in descending part, avoiding duplicate middle value.

# Visual for i = 4:

# Descending → 4321
# Ascending  → 234
# Result     → 4321234

print()

for i in range(1, n + 1):
    nums = list(range(i, 0, -1)) + list(range(2, i + 1))
    print(" " * (n - i) + "".join(map(str, nums)))

# Why this pattern is called palindrome?
# Because the numbers read the same left → right and right → left.

# Example: 543212345
```
