```python
name = "Lakshmi"
age = 25

print(f"My name is {name}, age is {age}")
print("My name is {}, age is {}".format(age, name))
print("My name is {1}, age is {0}".format(age, name))
print("My name is {name}, age is {age}".format(name="Lakshmi", age=25))

print("My name is %s, age is %d" % (name, age))
# %s → string
# %d → integer
# %f → float
```

```python
price = 96759.45674971235179
rate = 100000
discount = 0.75
name = 'Lakshmi'

print("Price: %.3f" % price)
print("Price: {:.2f}".format(price))
print(f"Price: {price:.5f}")
print(format(price, ".2f"))
print(format(price, ",.2f"))

print(f"Sales report for {name}: ₹{price:,.2f}")
print(f"Sum is {price + rate}")
print(f"{price = }")
print(f"{rate:,}")

print(f"{discount:.2%}")
print(f"{discount:06}") # 6(0 0 0 . 7 5)
```

```python
# alignment
text = "Lakshmi"

width = 30 if len(text) < 15 else 20
print(f"{text:>{width}}")

print(f"{'Left':<50}")
print(f"{'Center':^50}")
print(f"{'Right':>50}")

print("{:<50}".format("Left")) 
print("{:^50}".format("Center"))
print("{:>50}".format("Right"))
```

```python
name = "Lakshmi"
score = 92
print(f"{name} got {'Excellent' if score > 90 else 'Good'}")

num = 3.14159
precision = 2 if num > 3 else 4
print(f"{num:.{precision}f}")

text = "Hell"
width = 20 if len(text) < 5 else 5
print(f"{text:>{width}}")

print(f"{'A grade' if score>100 else 'B grade' if score>50 else 'C grade'}")
```

```python
marks = 75
salary = 45000

status_map = {True: "Pass", False: "Fail"}
print(f"Result: {status_map[marks >= 40]}")

result = f"Result: {'Pass' if marks >= 40 else 'Fail'}"
result = "Result: {}".format("Pass" if marks >= 40 else "Fail")
result = "Result: %s" % ("Pass" if marks >= 40 else "Fail")
print(result)

income = f"Category: {'High' if salary > 50000 else 'Medium' if salary > 30000 else 'Low'}"
print(f"{salary:,} ({'High' if salary > 50000 else 'Normal'})")
print(income)
```

```python
def get_grade(score):
    if score >= 90:
        return "A"
    elif score >= 60:
        return "B"
    else:
        return "C"

score = 85
print(f"Grade: {get_grade(score)}")
```

```python
customers = [{"name": "A", "spend": 200},
             {"name": "B", "spend": 2000}]

for c in customers:
    print(f"{c['name']} -> {'Premium' if c['spend'] > 1000 else 'Regular'}")

print(type(customers))
```

```python
# Use 'sep' → between values
# Use 'end' → after output
# Use 'print()' → for readability

# After every print there is (,end='\n') defaulty, invisible
# print("Hello",end='\n') = print("Hello")

a = "Lakshmi"
b = "Narasimha"

print(a, b, sep='\n', end='\n\n')
print(f"{a}\n{b}")
print(f"\n{a}\n{b}")
print()
print("Lakshmi ", end="")
print("Narasimha")
```
