Snake Case vs Pascal Case

```python
# | Feature   | snake_case                                       | PascalCase             |
# | --------- | ------------------------------------------------ | ---------------------- |
# | Words     | Lowercase                                        | Every word capitalized |
# | Separator | `_`                                              | None                   |
# | Example   | `employee_name`                                  | `EmployeeName`         |
# | Used for  | Variables, functions, methods, modules, packages | Classes, exceptions    |

class Employee:

    def calculate_salary(self):
        pass

# You can immediately identify:
    # Employee → Class
    # calculate_salary → Method

# Without reading the code further, the naming tells you what each identifier represents.
# Uses lowercase, snake_case for the class and PascalCase for the variables, method, which is opposite of PEP 8.

class SalesReport:

    def __init__(self, sales_data):
        self.sales_data = sales_data

    def calculate_total_sales(self):
        return self.sales_data["sales"].sum()

    def calculate_average_sales(self):
        return self.sales_data["sales"].mean()
    
# | Name                    | Type           | Convention                 |
# | ----------------------- | -------------- | -------------------------- |
# | SalesReport             | Class          | PascalCase                 |
# | init                    | Special method | Dunder (double underscore) |
# | sales_data              | Variable       | snake_case                 |
# | calculate_total_sales   | Method         | snake_case                 |
# | calculate_average_sales | Method         | snake_case                 |
```

## Real-Time Data Analyst Example (PEP 8)

```python
import pandas as pd


def clean_sales_data(file_path):
    """Load and clean sales data."""

    df = pd.read_csv(file_path)

    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_")
    )

    df["sales"] = df["sales"].fillna(0)
    df["city"] = df["city"].str.title()

    return df


def main():
    sales_df = clean_sales_data("sales.csv")
    print(sales_df.head())


if __name__ == "__main__":
    main()

# This example demonstrates several PEP 8 practices:
    # Imports at the top.
    # Two blank lines between top-level functions.
    # snake_case for function and variable names.
    # A docstring for the function.
    # Multi-line method chaining wrapped in parentheses.
    # Meaningful variable names.
    # The if __name__ == "__main__": entry point for executable scripts.
```
