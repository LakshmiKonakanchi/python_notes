```python
user_role = "admin"
allowed_roles = ["admin", "manager", "analyst"]

login_attempts = 5
max_attempts = 5

account_status = None
expected_status = None

flags_user = 0b1010   # binary flags
flags_required = 0b0010

# approch1
if (
    (user_role in allowed_roles)                     # membership (in)
    and (login_attempts < max_attempts)              # comparison
    and (account_status is expected_status)          # identity (is)
    and ((flags_user & flags_required) != 0)         # bitwise AND (&)
    and ((flags_user | flags_required) >= flags_user)# bitwise OR (|)
    and not (login_attempts == 0)                    # NOT
):
    print("Access Granted")
else:
    print("Access Denied")

# approch2
has_access = user_role in allowed_roles
valid_login = login_attempts < max_attempts
status_ok = account_status is None
flag_ok = (flags_user & flags_required) != 0

if has_access and valid_login and status_ok and flag_ok:
    print("Access Granted")
else:
    print("Access Denied")
```

```python
# Fraud Detection System (all operators)
transaction_amount = 12000
daily_limit = 10000
user_role = "guest"
blacklist = ["fraud_user", "blocked_user"]
username = "fraud_user"
flags = 0b1010   # bit flags

if (
    (transaction_amount > daily_limit)                   # comparison
    and (transaction_amount * 0.1 > 500)                 # arithmetic
    and (user_role in ["guest", "new_user"])             # membership
    and (username in blacklist)                          # membership
    and not (flags & 0b0001)                             # bitwise AND
):
    print("Fraud Alert 🚨")
else:
    print("Valid Customer ✅")
```

```python
# E-commerce (all operators)
cart_amount = 1500
items = ["laptop", "mouse"]
user_type = "premium"
discount_flag = 0b0011

if (
    (cart_amount >= 1000 and cart_amount <= 5000)   # comparison
    and ("laptop" in items)                         # membership
    and (user_type == "premium")                    # identity (also IS)
    and (discount_flag & 0b0001) != 0               # bitwise
    and any([cart_amount > 1200, len(items) > 2])   # any
    and all([cart_amount > 500, len(items) >= 1])   # all
):
    print("Discount Applied")
```

```python
# Senior-Level Cleaner Version
is_valid_amount = 1000 <= cart_amount <= 5000
has_laptop = "laptop" in items
is_premium = user_type == "premium"
has_discount = discount_flag & 0b0001
any_condition = any([cart_amount > 1200, len(items) > 2])
all_condition = all([cart_amount > 500, len(items) >= 1])

if all([is_valid_amount, has_laptop, is_premium, has_discount, any_condition, all_condition]):
    print("Discount Applied")
```
