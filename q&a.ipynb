## Shallow Copy vs Deep Copy

```python
# Shallow Copy vs Deep Copy

# Shallow Copy: Copies only the referance and the outer object only.
# Original changes because inner objects are shared.

import copy

original = [[1, 2], [3, 4]]
shallow = copy.copy(original)
shallow[0][0] = 100

print(original)

# Deep Copy: Creates completely independent copies.
# Original remains unchanged.

deep = copy.deepcopy(original)
deep[0][0] = 100

print(original)

# | Shallow Copy             | Deep Copy                 |
# | ------------------------ | ------------------------- |
# | Copies outer object only | Copies all nested objects |
# | Faster                   | Slower                    |
# | Less memory              | More memory               |
# | Shared references        | Independent references    |

# copy.copy()       # original changes
# copy.deepcopy()   # original not changed
```

## Iterator & Generator

```python
# Iterator
# An iterator is an object that allows sequential access to elements.
# An iterator is an object that implements __iter__() and __next__() methods.

nums = [10, 20, 30]
it = iter(nums)

print(next(it))
print(next(it))
print(next(it))

# Implements __iter__()
# Implements __next__()
# Produces one value at a time

# Used for memory-efficient processing.

####################################################

# Generator
# Generator produces values lazily.
# A generator is a function that uses yield to produce values one at a time without storing all values in memory.

def numbers():
    yield 1
    yield 2
    yield 3
gen = numbers()

print(next(gen))
print(next(gen))
print(next(gen))
print(next(gen))

##############################
gen = (x*x for x in range(5))
for value in gen:
    print(value)

# Why use Generators?
# Memory efficient, Faster for large datasets, Useful in ETL and Data Engineering pipelines

###################################

# | Iterator              | Generator      |
# | --------------------- | -------------- |
# | Class based           | Function based |
# | Uses   next           | Uses yield     |
# | More code             | Less code      |
# | Manual implementation | Automatic      |

# What happens internally when a generator function is called?
# When a generator function is called, Python does not execute the function immediately. 
# Instead, it returns a generator object. 
# Execution starts only when next() is called. 
# The generator maintains its state between calls and resumes execution from the last yield statement, making it memory efficient for processing large datasets.
```

## map() & filter() & reduce()

```python
# map()
# Applies a function to every element.

nums = [1, 2, 3, 4]
result = map(lambda x: x*2, nums)
print(list(result))

# Using Function

def square(x):
    return x*x
result = map(square, [1,2,3,4])
print(list(result))

#####################################################

# Filter()
# Filters data based on a condition.

nums = [1,2,3,4,5,6]
result = filter(lambda x: x % 2 == 0, nums)
print(list(result))

# Real Example
employees = [25000, 45000, 65000]

high_salary = filter(
    lambda x: x > 40000,
    employees
)
print(list(high_salary))

#######################################################

# reduce()
# Located in: from functools import reduce
# Used to reduce a collection into a single value.

from functools import reduce

# Sum Example
nums = [1,2,3,4]
result = reduce(
    lambda x, y: x + y,
    nums
)
print(result)

# Product Example
nums = [1,2,3,4]
result = reduce(
    lambda x, y: x * y,
    nums
)
print(result)

###############################################################

# map vs filter vs reduce

# | Function | Purpose   | Output       |
# | -------- | --------- | ------------ |
# | map()    | Transform | Collection   |
# | filter() | Select    | Collection   |
# | reduce() | Aggregate | Single Value |

from functools import reduce

nums = [1,2,3,4,5]

mapped = list(map(lambda x: x*2, nums))
filtered = list(filter(lambda x: x > 2, nums))
reduced = reduce(lambda x,y: x+y, nums)

print(mapped)
print(filtered)
print(reduced)
```
