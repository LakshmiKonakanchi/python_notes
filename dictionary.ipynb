```python
# N th highest salary
employees = {
    "emp1": {"name": "A", "salary": 50000},
    "emp2": {"name": "B", "salary": 70000},
    "emp3": {"name": "C", "salary": 90000},
    "emp4": {"name": "D", "salary": 50000},
    "emp5": {"name": "E", "salary": 10000},
    "emp6": {"name": "F", "salary": 90000}
}

salaries = {emp["salary"] for emp in employees.values()}  # unique salaries
second_highest = sorted(salaries)[-2]
print(second_highest)

Third_highest = sorted({e["salary"] for e in employees.values()})[-3]
result = [emp_id for emp_id, details in employees.items()
          if details["salary"] == Third_highest]
print(result)

salaries = sorted({e["salary"] for e in employees.values()}, reverse=True)
if len(salaries) >= 2:              # Edge Case method, strong approch
    print(salaries[1])
else:
    print("No second highest salary")

# Use set() → removes duplicates
# Use sorted() → easy logic
# Mention edge cases → strong answer

first = second = float('-inf')
for e in employees.values():
    sal = e["salary"]
    if sal > first:
        second = first
        first = sal
    elif first > sal > second:
        second = sal
print(second)
```

```python
text = 'tara ram pam pam tara ram pam pam ram para pam para'

# counter: for interview
from collections import Counter
freq = Counter(text.split())
print(freq)

# Pandas: Data Analyst Style
import pandas as pd
s = pd.Series(text.split())
print(s.value_counts())

# Manual Counting
words = text.split()
freq_dict = {}
for word in words:
    if word in freq_dict:
        freq_dict[word] += 1
    else:
        freq_dict[word] = 1
print(freq_dict)

# Pythonic way
dict_count = {word: text.split().count(word) for word in set(text.split())}
print(dict_count)
```

```python
d1 = {"a": 1, "b": 2, "d": 6}
d2 = {"c": 3, "c": 4, "d": 5}

d3 = {**d1, **d2}       # Merge two dictionaries  
print(d3)
# **d1 → unpacks all key-value pairs
# **d2 → adds more pairs
# Duplicate Key Case, right side wins

d1.update(d2)           # old method
print(d1)

d4 = d1 | d2            # using | Operator
print(d4)

d5 = {}
for d in (d1, d2):      # Using Loop
    for k, v in d.items():
        d5[k] = v
print(d5)
```

```python
# | Operation     | Code                      |
# | ------------- | ------------------------- |
# | Create dict   |   dict(zip(keys, values)) |
# | Unzip dict    |   zip(*d.items())         |
# | Loop pairs    |   for k,v in zip()        |
# | Handle uneven |   zip_longest()           |
```

```python
# zip() is used to combine multiple lists (or iterables) into pairs.

columns_keys = ["id", "name", "role"]
rows_values = [101, "Lakshmi", "Data Analyst"]

result = zip(columns_keys, rows_values)
print(dict(result), list(result), set(result), tuple(result), sep="\n")
# It can be used only once, after one full use → it becomes empty (exhausted)
print()
result = list(zip(columns_keys, rows_values))
print(dict(result))
print(list(result))
print(set(result))
print(tuple(result))
```

```python
# Unzip Dictionary (Reverse of zip)
# Convert dictionary → keys & values separately

user = {"id": 101, "name": "Lakshmi", "role": "Data Analyst"}

keys, values = zip(*user.items())
print(keys, values, sep="\n")

new_dict = dict(zip(keys, values))    # Convert Back to Dictionary
print(new_dict)

# user.items() → [('id',101), ('name','Lakshmi')]
# * → unpack
# zip reverses it
```

```python
keys = ["id", "name", "role", "place", "salary"]
values = [101, "Lakshmi", "Data Analyst"]

for k, v in zip(keys, values):
    print(k, v)                 # Extra values are ignored
print()


from itertools import zip_longest

keys = ["id", "name", "role", "place", "salary"]
values = [101, "Lakshmi", "Data Analyst"]

result = dict(zip_longest(keys, values, fillvalue=None))
print(result)                  # with extra values as None or any other
```

```python
# | Feature     | defaultdict             | Counter             |
# | ----------- | ----------------------- | ------------------- |
# | Purpose     | General default values  | Counting frequency  |
# | Flexibility | High                    | Limited to counting |
# | Default     | Custom (int, list, set) | Always int (0)      |

# Use Counter → when counting only
# Use defaultdict → when grouping or complex logic
```

```python
from collections import defaultdict

d = defaultdict(int)
d["a"] += 1
print(dict(d))

# int() → default value = 0
# So missing key → automatically becomes 0
# defaultdict automatically gives a default value if key does not exist.
```

```python
# Counting characters
from collections import defaultdict

text = "data science"
freq = defaultdict(int)

for ch in text:
    freq[ch] += 1
print(dict(freq))

for ch in text:
    if ch != " ":               # without space
        freq[ch] += 1
print(dict(freq))

text = text.replace(" ", "_")    # replace
print(text)
```

```python
# Grouping employees data by department

from collections import defaultdict

employees = [
    ("IT", "Lakshmi"),
    ("HR", "Ravi"),
    ("IT", "John"),
]

group = defaultdict(list)

for dept, name in employees:
    group[dept].append(name)

print(dict(group))
```

```python
# Using set (remove duplicates)

from collections import defaultdict

d = defaultdict(set)

d["a"].add(1)               # approch1
d["a"].add(1)
d["a"].add(2)

for v in [1, 1, 2]:         # approch2
    d["a"].add(v)

d["a"].update([1, 1, 2])    # approch3

print(dict(d))

d1 = defaultdict(set, {"a": {1, 2}})        # approch4
print(dict(d1))
```
