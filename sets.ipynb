```python
xyz = {0,1,2,3,4,5,6}
abc = set([5,6,7,8,9,0])
print(xyz | abc,    # Union: Combines both sets, removes duplicates
      xyz & abc,    # Intersection: Values present in both
      xyz - abc,    # Difference: Remove common values from xyz
      abc - xyz,    # Difference: Remove common values from abc
      xyz ^ abc,    # Symmetric Difference: All values except common ones
      sep='\n')

print('',
xyz.union(abc),
      xyz.intersection(abc),
      xyz.difference(abc),
      abc.difference(xyz),
      xyz.symmetric_difference(abc),
      sep='\n')
```

```python
A = {1, 2}
B = {1, 2, 3}
C = {1, 2, 3}
D = {1, 2}

print(A.issuperset(B)) # Does A contain ALL elements of B?
print(A.issubset(B))   #Are ALL elements of A present in B?

# | Concept  | Meaning            |
# | -------- | ------------------ |
# | subset   | small inside big   |
# | superset | big contains small |

E = {1, 2}
F = {1, 2, 3}

print(E <= F)  # subset
print(F >= E)  # superset
```

```python
G = {1, 2}
H = {1, 2, 3}
print(G < H)  # False (same elements)

# | Operator | Meaning         |
# | -------- | --------------- |
# | `<=`     | subset          |
# | `<`      | strict subset   |
# | `>=`     | superset        |
# | `>`      | strict superset |
```

```python
# Frozenset is just like a normal set, but immutable
# You cannot add, emove, update element
# frozenset is read-only
```

```python
fs1 = frozenset([1, 2, 3])
fs2 = frozenset([3, 4, 5])

print(fs1.union(fs2))    
print(fs1.intersection(fs2)) 
print(fs1.difference(fs2)) 
print(type(fs1))
```

```python
# | Type      | Syntax             |
# | --------- | ------------------ |
# | list      | [1,2,3]            |
# | set       | {1,2,3}            |
# | frozenset | frozenset([1,2,3]) |

# frozenset has no literal syntax like {}, []
# frozenset() expects only one iterable
# fs1 = frozenset{1, 2, 3}            # error
fs2 = frozenset[1, 2, 3]            # wrong
fs3 = frozenset([1, 2, 3])          # correct

print(type(fs2), type(fs3))

```

```python
("A", "B") != ("B", "A") # Normal tuple fails, but in frozonsets
edges = {
    frozenset(["A", "B"]),
    frozenset(["B", "A"]),
    frozenset(["A", "C"])
}
print(edges)
print(type(edges))

# edges1 = frozenset(["A", "B"],["B", "A"],["A", "C"])
# edges1 is error: frozenset() takes only ONE iterable, not multiple arguments

# If you want one frozenset containing multiple pairs:
edges2 = frozenset([
    frozenset(["A", "B"]),
    frozenset(["B", "A"]),
    frozenset(["A", "C"])
])
print(edges2)
```

```python
features = [
    frozenset(["age", "salary"]),
    frozenset(["salary", "age"]),
    frozenset(["experience", "salary"])
]
print(set(features)) #Removes duplicate combinations

unique_features = set(frozenset(i) for i in features)
print(unique_features)
```

```python
cache = {}

def get_result(items):
    key = frozenset(items)
    
    if key in cache:
        return cache[key]
    
    result = sum(items)  # expensive operation
    cache[key] = result
    
    return result

print(get_result([1,2,3]))
print(get_result([3,2,1]))  # reused!
# Avoid recomputation
# Order-independent caching
```
